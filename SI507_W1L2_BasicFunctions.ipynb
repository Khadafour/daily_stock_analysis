## Part I: Understanding the Problem - Why Functions?

Look at this sales tracking code. Can you figure out what it does?

````python
# Our data structures
SalesPeople = {
    '1001': {'last_name': 'C', 'first_name': 'Alan', 'uniqname': 'AlanC'},
    '1002': {'last_name': 'R', 'first_name': 'Keisha', 'uniqname': 'RKeisha'},
    '1003': {'last_name': 'W', 'first_name': 'Tanya', 'uniqname': 'WTanya'},
    '1004': {'last_name': 'D', 'first_name': 'Henry', 'uniqname': 'DHenry'}
}
sales = {
    '1001': [190, 188, 175, 195],
    '1002': [192, 199, 288, 140],
    '1003': [195, 188, 182, 110],
    '1004': [99, 292, 94, 298]
}

# Finding the salesperson with highest average
max = 0
max_id = -1
for id in sales:
    sum = 0
    num_sales = 0
    for s in sales[id]:
        sum = sum + s
        num_sales += 1
    avg = sum/num_sales
    if avg > max:
        max = avg
        max_id = id
print(SalesPeople[max_id])
````

### A Critical Bug: The Shadow Problem

Before we even talk about functions, this code has a dangerous bug. Run this:

````python
# After running the sales code above, try:
print(max([1, 2, 3]))  # TypeError: 'float' object is not callable
````

What happened? We accidentally **shadowed** Python's built-in functions!

````python
max = 0    # This replaces Python's max() function with the number 0
sum = 0    # This replaces Python's sum() function with the number 0
id = '1001'  # This replaces Python's id() function with a string

# Now we can't use these built-in functions anymore in our session!
````

**First Rule of Functions:** Never use Python built-in names as variable names.

Common built-ins to avoid: `id`, `sum`, `max`, `min`, `list`, `dict`, `str`, `int`, `float`, `input`, `print`, `type`, `len`, `all`, `any`, `zip`, `map`, `filter`

Let's fix our code:

````python
# FIXED: Using descriptive names that don't shadow built-ins
best_avg = 0
best_id = -1
for emp_id in sales:
    total = 0
    num_sales = 0
    for s in sales[emp_id]:
        total = total + s
        num_sales += 1
    avg = total/num_sales
    if avg > best_avg:
        best_avg = avg
        best_id = emp_id
print(SalesPeople[best_id])
````

## Part III: The Power of Reusability - DRY vs WET

Let's expand our program to also show everyone's averages:

````python
def find_salesperson_with_highest_average(SalesPeople, sales):
    best_avg = 0
    best_id = -1
    for emp_id in sales:
        total = 0
        num_sales = 0
        for s in sales[emp_id]:
            total += s
            num_sales += 1
        avg = total/num_sales
        if avg > best_avg:
            best_avg = avg
            best_id = emp_id
    return SalesPeople[best_id]

def compute_final_sales(sales):
    salesperson_sales = {}
    for emp_id in sales:
        total = 0
        num_sales = 0
        for s in sales[emp_id]:
            total += s
            num_sales += 1
        salesperson_sales[emp_id] = total/num_sales
    return salesperson_sales

def print_final_sales(SalesPeople, final_sales):
    for emp_id in final_sales:
        salesperson = SalesPeople[emp_id]
        print(salesperson['first_name'], salesperson['last_name'], 'has', final_sales[emp_id])

# Using our functions
top_salesperson = find_salesperson_with_highest_average(SalesPeople, sales)
print("The top salesperson is", top_salesperson['first_name'], top_salesperson['last_name'])

final_sales = compute_final_sales(sales)
print_final_sales(SalesPeople, final_sales)
````

### Identifying Repeated Code

Look carefully at the two functions above. Do you see repeated code? This section appears twice:

````python
total = 0
num_sales = 0
for s in sales[emp_id]:
    total += s
    num_sales += 1
avg = total/num_sales
````

This violates the **DRY Principle** (Don't Repeat Yourself). When we write the same code multiple times (WET - Write Every Time), we:
- Risk typos and inconsistencies
- Make changes harder (need to update multiple places)
- Make code longer and harder to maintain

### Refactoring: Extracting the Common Pattern

*Refactoring* means restructuring code without changing its behavior. Let's extract the repeated averaging logic:

````python
def compute_average(num_list):
    """Calculate the average of a list of numbers."""
    total = 0
    num_items = 0
    for n in num_list:
        total += n
        num_items += 1
    return total/num_items

# Now our other functions become much simpler:
def find_salesperson_with_highest_average(SalesPeople, sales):
    best_avg = 0
    best_id = -1
    for emp_id in sales:
        avg = compute_average(sales[emp_id])  # Using our new function!
        if avg > best_avg:
            best_avg = avg
            best_id = emp_id
    return SalesPeople[best_id]

def compute_final_sales(sales):
    salesperson_sales = {}
    for emp_id in sales:
        salesperson_sales[emp_id] = compute_average(sales[emp_id])  # Reused!
    return salesperson_sales
````

Now if we need to change how averaging works (maybe handle empty lists), we only change it in ONE place.

## Part IV: Making Functions Safe - Handling Edge Cases

What happens if we call our function with an empty list?

````python
compute_average([])  # ZeroDivisionError: division by zero
````

Let's make it safer with doctests (executable documentation):

````python
def compute_average(numbers):
    """
    Calculate the average of a list of numbers.

    Args:
        numbers: A list of numeric values

    Returns:
        float: The arithmetic mean

    Raises:
        ValueError: If the list is empty

    Examples:
        >>> compute_average([10, 20, 30])
        20.0
        >>> compute_average([1])
        1.0
        >>> compute_average([100, 200])
        150.0
        >>> compute_average([])
        Traceback (most recent call last):
            ...
        ValueError: Cannot compute average of empty sequence
    """
    if not numbers:  # Empty list check
        raise ValueError("Cannot compute average of empty sequence")

    total = 0
    count = 0
    for num in numbers:
        total += num
        count += 1

    return total / count
````

### Running Doctests

Those `>>>` lines aren't just documentation - they're tests we can run!

````python
import doctest

# Test just one function
doctest.run_docstring_examples(compute_average, globals(), verbose=True)

# Test everything in the module
doctest.testmod(verbose=True)
````

Output shows each test passing or failing. Doctests serve three purposes:
1. **Documentation** - Show users how to use the function
2. **Testing** - Verify the function works
3. **Examples** - Provide copy-paste examples

## Part VI: Understanding Scope - Local vs Global

All variables in Python aren't accessible everywhere. The **scope** determines where you can access a variable.

### Basic Scope Example

````python
var1 = 66  # Global scope

def func1():
    print("inside function", var1)  # Can access global

func1()  # Prints: inside function 66
print("outside function", var1)  # Prints: outside function 66
````

### Local Variables Override Global

````python
var2 = 77  # Global scope

def func2():
    var2 = 88  # Local variable (shadows global)
    print("inside function", var2)

func2()  # Prints: inside function 88
print("outside function", var2)  # Prints: outside function 77 (unchanged!)
````

The local `var2` is a completely different variable that happens to have the same name. When the function ends, the local variable disappears.

### The LEGB Rule

Python looks for variables in this order:
1. **L**ocal - Inside current function
2. **E**nclosing - In enclosing function (nested functions)
3. **G**lobal - Module level
4. **B**uilt-in - Pre-defined names like `print`

````python
x = "global"

def outer():
    x = "enclosing"

    def inner():
        x = "local"
        print(x)  # Prints "local"

    inner()
    print(x)  # Prints "enclosing"

outer()
print(x)  # Prints "global"
````

### Using `nonlocal` (Advanced)

Sometimes you need to modify an enclosing scope variable:

````python
def make_counter():
    count = 0

    def increment():
        nonlocal count  # Modify enclosing variable
        count += 1
        return count

    return increment

counter = make_counter()
print(counter())  # 1
print(counter())  # 2
print(counter())  # 3
````

Without `nonlocal`, Python would create a new local variable instead of modifying the enclosing one.

## Part VII: Mutability and Functions - Critical Concept

Remember: Numbers, strings, and tuples are **immutable**. Lists and dictionaries are **mutable**.

### Immutable Arguments Are Safe

````python
def func6(var6):
    var6 += " world"
    print(var6, "from func6")

var6 = "hello"
func6(var6)  # Prints: hello world from func6
print(var6, "from main")  # Prints: hello from main (unchanged!)
````

### Mutable Arguments Can Be Dangerous

````python
def dict_update(local_dict):
    local_dict['g'] = 'grapefruit'  # Modifies the original!
    local_dict['n'] = 'nectarine'

global_dict = {'g': 'grape', 'p': 'plum'}
print(global_dict['g'])  # grape

dict_update(global_dict)

print(global_dict['g'])  # grapefruit (changed!)
print(global_dict['n'])  # nectarine (added!)
````

Why? `local_dict` isn't a copy - it's a reference to the same dictionary object!

### The Mutable Default Trap

One of Python's most famous gotchas:

````python
def add_item_bad(item, basket=[]):
    basket.append(item)
    return basket

result1 = add_item_bad('apple')
print(result1)  # ['apple']
````

````python
result2 = add_item_bad('banana')
print(result2)  # ['apple', 'banana'] - Wait, what?!
````

````python
# DANGEROUS: Mutable default argument
def add_item_bad(item, basket=[]):
    basket.append(item)
    return basket

result1 = add_item_bad('apple')
print(result1)  # ['apple']

result2 = add_item_bad('banana')
print(result2)  # ['apple', 'banana'] - Wait, what?!

# The default list is created ONCE when function is defined!
````

**Safe Pattern:**

````python
def add_item_safe(item, basket=None):
    if basket is None:
        basket = []  # Create new list each call
    basket.append(item)
    return basket

result1 = add_item_safe('apple')
print(result1)  # ['apple']

result2 = add_item_safe('banana')
print(result2)  # ['banana'] - Correct!
````

## Part VIII: The Aliasing Trap with Multiplication

````python
# TRAP: Multiplying mutable objects
dict_list = [{}] * 2
print(dict_list)  # [{}, {}] - looks fine

dict_list[0]['key'] = 'value'
print(dict_list)  # [{'key': 'value'}, {'key': 'value'}] - Both changed!

# Why? Both elements reference the SAME dictionary!
````

**Safe approach using list comprehension:**

````python
dict_list = [{} for _ in range(2)]  # Create separate dictionaries
dict_list[0]['key'] = 'value'
print(dict_list)  # [{'key': 'value'}, {}] - Only first changed
````

The `_` is Python convention for "unused variable" - we don't care about the loop value.

## Part IX: Lambda Functions and Functional Tools

### Lambda: Anonymous Functions

Sometimes we need a small function just once. Lambda provides a shorthand:

````python
# Traditional function
def double(x):
    return x * 2

# Lambda equivalent
double = lambda x: x * 2

# Lambda syntax: lambda arguments: expression
print(double(5))  # 10
````

### Using `max()` with `key` Parameter

Instead of manually finding the maximum, use Python's built-in with a key function:

````python
# Our refactored code can be even simpler!

# Old way (manual loop):
best_avg = 0
best_id = -1
for emp_id in sales:
    avg = compute_average(sales[emp_id])
    if avg > best_avg:
        best_avg = avg
        best_id = emp_id

# New way (functional):
best_id = max(sales.keys(), key=lambda emp_id: compute_average(sales[emp_id]))

print(f"Top performer: {SalesPeople[best_id]['first_name']}")
````

The `key` parameter tells `max()` what to compare. For each `emp_id`, it calls the lambda function to get the comparison value (the average).

## Part X: Advanced Parameter Patterns

### Variable Arguments: *args

````python
def compute_sum(*args):
    """Sum any number of arguments."""
    total = 0
    for num in args:
        total += num
    return total

print(compute_sum(1, 2, 3))  # 6
print(compute_sum(10))  # 10
print(compute_sum(1, 2, 3, 4, 5))  # 15
````

### A Ramalho-style Deep Dive into *args and **kwargs

In Python, functions are not rigid molds; they are flexible contracts. The `*args` and `**kwargs` syntax embodies this philosophy of flexibility.

- `*args` gathers positional arguments into a tuple. Think of it as saying: "Give me however many unnamed arguments you like, and I will bundle them neatly." Inside the function, you can iterate over this tuple, just as if you were handed a list.

- `**kwargs` collects keyword arguments into a dictionary. This allows a function to accept named arguments that were not explicitly declared in its signature. It is as though you are saying: "If the caller has more to say, I am listening." Each extra keyword becomes a key in the dictionary, mapping to its value.

This design is quintessential Pythonic elegance: instead of closing doors, functions open themselves to extension.

A practical idiom: functions that forward arguments to another layer often use `*args` and `**kwargs`. For instance, decorators or wrapper functions do not know in advance what exact parameters the wrapped function expects. By using these two constructs, they can forward everything faithfully:

```python
def wrapper(func):
    def inner(*args, **kwargs):
        print("Calling", func.__name__)
        return func(*args, **kwargs)  # pass everything along
    return inner
```

Notice how this simple idiom captures Python’s ethos: trust the caller, accept variety, and forward it with grace.


````python
def create_profile(**kwargs):
    """Create profile from keyword arguments."""
    return kwargs

profile = create_profile(name='Alice', age=30, city='NYC')
print(profile)  # {'name': 'Alice', 'age': 30, 'city': 'NYC'}
````

### Keyword-Only Parameters (Python 3+)

Force certain parameters to be named:

````python
def fetch_data(url, *, timeout=30, retries=3):
    """The * forces timeout and retries to be keyword-only."""
    print(f"Fetching {url}, timeout={timeout}, retries={retries}")

fetch_data('http://api.com', timeout=60)  # OK
# fetch_data('http://api.com', 60)  # ERROR! Must use keyword
````

## Part XI: Introduction to Decorators

A decorator is a function that wraps another function to add behavior:

````python
import time

def timer(func):
    """Decorator to time function execution."""
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {end-start:.4f} seconds")
        return result
    return wrapper

@timer
def slow_function(n):
    total = 0
    for i in range(n):
        total += i
    return total

result = slow_function(1000000)
# Prints: slow_function took 0.0421 seconds
````

The `@timer` syntax is equivalent to:

````python
slow_function = timer(slow_function)
````

## Part XII: Functional vs In-Place Operations

Python has two styles for many operations:

````python
# Functional - returns new object
numbers = [3, 1, 4]
sorted_nums = sorted(numbers)
print(numbers)      # [3, 1, 4] - unchanged
print(sorted_nums)  # [1, 3, 4] - new list

# In-place - modifies original
numbers.sort()
print(numbers)  # [1, 3, 4] - changed!
````

Strings are immutable, so only have functional methods:

````python
name = "hello"
upper = name.upper()  # Returns new string
print(name)   # "hello" - unchanged
print(upper)  # "HELLO"
````

## Complete Refactored Solution

Bringing it all together with everything we've learned:

````python
import doctest

def safe_average(numbers):
    """
    Calculate average with error handling.

    >>> safe_average([10, 20, 30])
    20.0
    >>> safe_average([])
    Traceback (most recent call last):
        ...
    ValueError: Cannot average empty sequence
    """
    if not numbers:
        raise ValueError("Cannot average empty sequence")
    return sum(numbers) / len(numbers)

# Analyze sales functionally
def analyze_sales(salespeople, sales_data):
    """Complete sales analysis using functional approach."""

    # Get all averages
    averages = {
        emp_id: safe_average(sales)
        for emp_id, sales in sales_data.items()
    }

    # Find top performer
    top_id = max(averages, key=averages.get)

    # Sort all by performance
    ranking = sorted(averages.items(),
                    key=lambda x: x[1],
                    reverse=True)

    return {
        'top': salespeople[top_id],
        'ranking': ranking,
        'averages': averages
    }

# Run analysis
results = analyze_sales(SalesPeople, sales)
print(f"Top: {results['top']['first_name']} {results['top']['last_name']}")
for emp_id, avg in results['ranking']:
    person = SalesPeople[emp_id]
    print(f"{person['first_name']:10} {avg:.2f}")

# Test our functions
if __name__ == "__main__":
    doctest.testmod(verbose=True)
````
