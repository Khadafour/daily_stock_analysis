## Part I: Input and the Python Type System

*"In Python, everything is an object, but not all objects are created equal."*

When we accept user input, Python makes a deliberate choice: everything comes in as a string. This is both a blessing and a curse. Let's explore why.

**Design note — Explicit beats implicit.**

`input()` always returns `str`. If you want numbers, dates, or domain-specific objects, *you* are responsible for the transformation. This explicit step is not busywork; it documents intent.

````python

# The fundamental truth about input()
user_data = input("Tell me anything: ")
print(f"Type: {type(user_data)}, Value: {user_data!r}")

# Even if the user types: 42
# We receive: '42' (a str object)
````

### The Type Hierarchy: A Mental Model

Before we dive deeper, let's establish a mental model. In Python's type system for basic input handling:

```
object (everything inherits from this)
    ├── str   (most general, accepts anything typed)
    ├── float (accepts numeric values with decimals)
    └── int   (most restrictive, whole numbers only)
```
This hierarchy is not about inheritance between `str`, `float`, and `int`; rather it is about *constraints on conversion*. Converting from `str` to `int` is strict; converting from `str` to `float` is more permissive. When in doubt, be explicit.

````python

s = "42.5"
print(float(s))   # 42.5 works

try:
    print(int(s))  # ValueError: '42.5' is not an int literal
except ValueError as e:
    print("Cannot convert directly to int:", e)

# If you accept truncation, you can do:
print(int(float(s)))  # 42 (note: potential data loss)
````

## Part II: Introduction to Exceptions — Handling Errors Gracefully

Instead of letting our program crash, Python provides **exceptions**, a control-flow mechanism that allows us to handle failures close to their source.
The guiding mindset is **EAFP**: *Easier to Ask Forgiveness than Permission*. This is contrast to other programming languages which use LBYL or *Look Before You Leap*. Rather than pre-validate every path, perform the action and handle the exception if it fails.

A useful side-effect of EAFP is that your code mirrors the *happy path*: the main logic remains uncluttered, and the exceptional path is explicit and local.

````python

# Basic structure:
try:
    # Code that might fail
    risky_operation()
except SomeException:
    # What to do if it fails
    handle_error()
````

**Common exceptions with input**

- `ValueError`: Wrong type conversion (e.g., `int("abc")`).
- `TypeError`: Operation applied to an object of inappropriate type.
- `KeyboardInterrupt`: User pressed Ctrl+C to stop the program.
- `EOFError`: End of input (e.g., Ctrl+D / Ctrl+Z).

In real programs, catch the *specific* exception you anticipate and can handle responsibly.

````python

try:
    age = int(input("How old are you? "))
    print("In 5 years you will be:", age + 5)
except ValueError:
    print("Please enter a whole number, like 25.")
````

## Part III: Defensive Programming for User Input

A defensive input loop is a compact pattern that pays for itself quickly. It communicates requirements to the user, validates ranges, and recovers from errors—without obscuring the main logic.

````python

while True:
    try:
        age = int(input("Enter your age: "))
        if not 0 <= age <= 120:
            print("Please enter a realistic age between 0 and 120.")
            continue
        break
    except ValueError:
        print("Please enter a whole number.")
    except KeyboardInterrupt:
        print("\nExiting. Goodbye!")
        break
````

### Multiple Inputs with Simple Rules

````python

# Collect a few fields procedurally
profile = {}

# Name
while True:
    name = input("Name: ").strip()
    if name and name.replace(" ", "").isalpha():
        profile["name"] = name
        break
    print("Use only letters and spaces.")

# Email
while True:
    email = input("Email: ").strip()
    if "@" in email and "." in email:
        profile["email"] = email
        break
    print("Please enter a valid email address.")

# Age
while True:
    try:
        age = int(input("Age: "))
        if 0 <= age <= 120:
            profile["age"] = age
            break
        print("Age must be 0–120.")
    except ValueError:
        print("Whole numbers only, please.")

print("\nSaved profile:")
print(profile)
````

## Part IV: The Gas Tank Example

This is the same input-loop pattern specialized for a simple decision. A small program like this benefits from clarity: a straightforward loop, a clear range, and three explicit outcomes.

````python

print("Gas Tank Refueling Light")
print("/\\" * 10)

while True:
    try:
        gas_level = int(input("What percent full is the gas tank? "))
        if not 0 <= gas_level <= 100:
            print("Please enter a percentage between 0 and 100")
            continue
        break
    except ValueError:
        print("Please enter a number without the % sign (e.g., 50)")
    except KeyboardInterrupt:
        print("\nProgram cancelled.")
        break

if gas_level > 50:
    print("We have enough gas")
elif gas_level > 10:
    print("We need to get more gas soon")
else:
    print("Please stop at the very next gas station!")
````

## Part V: Assertions vs Exceptions

**Assertions** are for the programmer: they document invariants and catch logic errors while developing or testing.  
**Exceptions** are for the user: they express recoverable problems in production code.

Two rules worth repeating:
1. Assertions can be disabled; do not rely on them for user validation.
2. Catch exceptions *specifically*, and only when you can handle them meaningfully.

````python

# Assertion example (development-time check)
nums = []
assert len(nums) > 0, "Cannot compute average of empty list"
# avg = sum(nums) / len(nums)
````

````python

# Exception-driven validation for user input
text = input("Age: ")
if not text.isdigit():
    raise ValueError("Age must be a number")
age = int(text)
````

## Part VI: Pitfalls and Anti-Patterns

Here are a few examples of good and bad practices.

BAD!!! Bare execept that catches everything. it hides the bugs and it blocks Ctrl+C

````python

try:
    value = int(input())
except:
    pass
````

Much Better: Catch a specific exception

````python

try:
    value = int(input())
except ValueError:
    print("Invalid number")
````

BAD! Using assert for user validation (can be disabled; not user-friendly)


````python

age = input("Age: ")
assert age.isdigit()
````

BETTER: Explicit checks

````python


age = input("Age: ")
if not age.isdigit():
    raise ValueError("Age must be a number")
````
